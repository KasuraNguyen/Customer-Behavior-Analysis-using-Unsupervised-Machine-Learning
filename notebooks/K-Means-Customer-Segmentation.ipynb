```python
# Chuyển Purchase_Amount từ chuỗi sang số
df['Purchase_Amount'] = pd.to_numeric(
    df['Purchase_Amount']
        .astype('string')
        .str.replace('$', '', regex=False)
        .str.replace(',', '', regex=False),
    errors='coerce'
)

print("Kiểu dữ liệu:", df['Purchase_Amount'].dtype)
print("Số giá trị thiếu:", df['Purchase_Amount'].isnull().sum())
print(df['Purchase_Amount'].head(10))
```

```python
#Kiem tra cac cot duoc lua chon hop li
cot_so = [
    'Age',
    'Purchase_Amount',
    'Frequency_of_Purchase',
    'Brand_Loyalty',
    'Product_Rating',
    'Customer_Satisfaction',
    'Time_to_Decision'
]

df[cot_so].describe().T
```

```python
#kiem tra gia tri bi thieu trong 7 cot tren
print(df[cot_so].isnull().sum())
```

```python
from sklearn.preprocessing import StandardScaler

X = df[cot_so].copy() #lay cac cot da chon trc do

#chuan hoa du lieu
scaler = StandardScaler() #tao cong cu chuan hoa
X_scaled = scaler.fit_transform(X) #fit_trans: tinh trung binh, do lech chuan cua tung cot roi bien doi du lieu

print("Kích thước dữ liệu ban dầu: ", X.shape)
print("Kích thước dữ liệu sau chuẩn hóa: ", X_scaled.shape)
#test
print("\n 5 dòng đầu tiên sau chuẩn hóa: ")
print(X_scaled[:5])
```

```python
#Tìm số cụm K phù hợp bằng phương pháp Elbow
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Lưu tổng khoảng cách bình phương của các điểm đến tâm cụm
inertias = []

# Thử số cụm K từ 1 đến 10
K_range = range(1, 11)

# Lần lượt chạy K-Means với từng giá trị K
for k in K_range:

    # Tạo mô hình K-Means với k cụm
    model = KMeans(
        n_clusters=k,   # Số cụm muốn chia
        random_state=42, # Cố định ngẫu nhiên để dễ tái lập kết quả
        n_init=10        # Thử 10 lần khởi tạo tâm cụm
    )

    # Huấn luyện mô hình trên dữ liệu đã chuẩn hóa
    model.fit(X_scaled)

    # Lưu giá trị Inertia để so sánh các giá trị K
    inertias.append(model.inertia_)

# Tạo khung vẽ biểu đồ
plt.figure(figsize=(8, 5))

# Vẽ đường biểu diễn K và Inertia
plt.plot(K_range, inertias, marker='o')

# Đặt tên cho trục ngang và trục dọc
plt.xlabel('Số cụm K')
plt.ylabel('Inertia')

# Đặt tiêu đề biểu đồ
plt.title('Elbow Method')

# Hiển thị các giá trị K trên trục ngang
plt.xticks(list(K_range))

# Thêm lưới để dễ đọc biểu đồ
plt.grid(True)

# Hiển thị biểu đồ
plt.show()
```

```python
# In giá trị Inertia tương ứng với từng số cụm K
for k, inertia in zip(K_range, inertias):
    print(f"K = {k}: Inertia = {inertia:.2f}")
```

```python
#Do từ K=3 trở đi, Inertia giảm nhỏ dần nên vẫn chưa khẳng định được K = 3 là tốt nhất. Ta thử chia thành 3 cụm

#===Huấn luyện mô hình K-Means với 3 cụm

from sklearn.cluster import KMeans
kmeans = KMeans(
    n_clusters =3,
    random_state = 42,
    n_init = 10
)

kmeans.fit(X_scaled)
df['Cluster'] = kmeans.labels_ #gán nhãn cụm cho từng khách

#Đếm số khách của từng cụm
print(df['Cluster'].value_counts().sort_index())
```

```python
#Xem trung bình từng cụm
bpt = df.groupby('Cluster')[cot_so].mean()
bpt = bpt.round(2)
print(bpt)
```

```python
#Vẽ biểu đồ để trực quan hóa 3 cụm (Quan trọng)

import matplotlib.pyplot as plt

plt.figure(figsize = (8,5)) #Khung vẽ biểu đồ

plt.scatter(
    df['Purchase_Amount'],   # Trục X: giá trị mua hàng
    df['Product_Rating'],    # Trục Y: điểm đánh giá sản phẩm
    c = df['Cluster'],       # Màu sắc: dựa theo nhãn cụm (Cluster) mà K-Means đã gán
    cmap ='viridis',         # Bảng màu: dùng colormap "viridis" (chuyển từ xanh sang vàng)
    alpha = 0.7              # Độ trong suốt: 0.7 nghĩa là hơi mờ để dễ nhìn chồng lấn
)

#Đặt tên
plt.xlabel('Purchase_Amount')
plt.ylabel('Product_Rating')

#Tên biểu đồ
plt.title('Customer Cluster: Purchase Amount & Product Rating')

#Chú thích màu theo cụm
plt.colorbar(label = 'Cluster')

#Hiển thị
plt.show()
```
